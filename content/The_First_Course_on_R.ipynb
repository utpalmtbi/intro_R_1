## 3. Arithmetic & Relational Operators — Fold Change & Cutoffs

**The problem:** You have raw Ct values for control and treated samples.
How do you calculate fold change? And which genes are truly upregulated?

**The solution:** Arithmetic operators for math; relational operators for comparisons.

### Arithmetic: Calculating 2^(-ΔΔCt)

```r
# Ct values for a reference gene (e.g., GAPDH) and target genes
ct_ref <- 22.1       # GAPDH Ct
ct_target <- 28.5    # Target gene Ct in treated sample
ct_control <- 30.0   # Target gene Ct in control sample

# ΔCt for treated and control
dct_treated <- ct_target - ct_ref
dct_control <- ct_control - ct_ref

# ΔΔCt
ddct <- dct_treated - dct_control

# Fold change = 2^(-ΔΔCt)
fold_change <- 2 ^ (-ddct)
fold_change
```

### Common arithmetic operators

```r
a <- 10
b <- 3

a + b    # addition
a - b    # subtraction
a * b    # multiplication
a / b    # division
a ^ b    # exponentiation (10³ = 1000)
a %% b   # modulo (remainder: 10/3 = 3 remainder 1)
```

### Relational: Which genes pass the cutoff?

```r
fold_changes <- c(1.2, 4.5, 0.3, 8.1, 2.0, 0.9)
gene_names <- c("BRCA1", "TP53", "EGFR", "MYC", "KRAS", "PTEN")

# Which genes have fold change > 2? (upregulated)
fold_changes > 2

# Get the actual gene names that pass the cutoff
gene_names[fold_changes > 2]
```

### All relational operators

```r
values <- c(1, 5, 10, 15, 20)

values > 10    # greater than
values < 10    # less than
values == 10   # equal to (==, not =)
values != 10   # not equal to
values >= 10   # greater than or equal
values <= 10   # less than or equal
```

## 4. Logical Operators — Combining Conditions

**The problem:** You want genes that are BOTH significantly upregulated (fold change > 2)
AND have high expression (mean Ct < 30). How do you combine conditions?

**The solution:** Logical operators — `&` (AND), `|` (OR), `!` (NOT).

```r
fold_changes <- c(1.2, 4.5, 0.3, 8.1, 2.0, 0.9)
ct_values <- c(28.5, 31.2, 27.8, 35.1, 29.6, 33.0)
gene_names <- c("BRCA1", "TP53", "EGFR", "MYC", "KRAS", "PTEN")

# Fold change > 2 AND Ct < 30
condition <- fold_changes > 2 & ct_values < 30
condition

# Which genes satisfy both?
gene_names[condition]
```

### OR and NOT in action

```r
# OR: fold change > 4 OR Ct < 28
gene_names[fold_changes > 4 | ct_values < 28]

# NOT: genes that are NOT upregulated
gene_names[!(fold_changes > 2)]
```

## 9. Decision Making — Classifying Results

**The problem:** Based on fold change and p-value, you want to label each gene as:
- "Upregulated" (FC > 2, p < 0.05)
- "Downregulated" (FC < 0.5, p < 0.05)
- "Not significant" (everything else)

**The solution:** `if-else` statements — your code makes decisions.

```r
# Classify one gene
fc <- 4.5
pval <- 0.003

if (fc > 2 & pval < 0.05) {
    print("Upregulated")
} else if (fc < 0.5 & pval < 0.05) {
    print("Downregulated")
} else {
    print("Not significant")
}
```

### Apply if-else to each row of your data frame

```r
# Loop through each row using a for loop (we'll learn this properly next)
# For now, see how decision-making works on real data

for (i in 1:nrow(results)) {
    fc <- results$FoldChange[i]
    pval <- results$PValue[i]
    
    if (fc > 2 & pval < 0.05) {
        results$Status[i] <- "Upregulated"
    } else if (fc < 0.5 & pval < 0.05) {
        results$Status[i] <- "Downregulated"
    } else {
        results$Status[i] <- "Not significant"
    }
}

results
```

## 11. Functions — Build Your Own Tools

**The problem:** You keep calculating 2^(-ΔΔCt) the same way for every experiment.
Copy-paste is error-prone. Can you package this calculation into a reusable tool?

**The solution:** Write your own **function**.

```r
# Write a function to compute fold change from Ct values
fold_change <- function(ct_target, ct_control, ct_ref) {
    dct_treated <- ct_target - ct_ref
    dct_control <- ct_control - ct_ref
    ddct <- dct_treated - dct_control
    fc <- 2 ^ (-ddct)
    return(fc)
}

# Use it
fold_change(ct_target = 28.5, ct_control = 30.0, ct_ref = 22.1)
```

### Functions with default arguments

```r
# Default threshold for significance
classify_gene <- function(fc, pval, fc_cutoff = 2.0, pval_cutoff = 0.05) {
    if (fc > fc_cutoff & pval < pval_cutoff) {
        return("Upregulated")
    } else if (fc < 1/fc_cutoff & pval < pval_cutoff) {
        return("Downregulated")
    } else {
        return("Not significant")
    }
}

# Call with defaults
classify_gene(4.5, 0.003)

# Call with custom cutoff
classify_gene(4.5, 0.003, fc_cutoff = 3.0)
```

### Returning multiple values

```r
# A function that returns summary statistics as a list
gene_summary <- function(values) {
    list(
        mean = mean(values),
        sd = sd(values),
        min = min(values),
        max = max(values)
    )
}

gene_summary(c(1.2, 4.5, 0.3, 8.1, 2.0, 0.9))
```

## 13. Regular Expressions — Finding Patterns in Sequences

**The problem:** You need to find restriction enzyme cut sites in a DNA sequence.
EcoRI cuts at `GAATTC`. How many sites? Where are they?

**The solution:** **Regular expressions** — a mini-language for pattern matching in text.

```r
# Our DNA sequence
seq <- "ATGAATTCGGCCGAATTCGTTAAGAATTCCTG"

# Find the EcoRI sites (GAATTC)
grep("GAATTC", seq, value = TRUE)
```

### Quantifiers — flexible pattern matching

```r
# Sequences to search
sequences <- c("ATGC", "AATGC", "AAATGC", "AAAATGC")

# 'A+' = one or more A's
grep("A+TGC", sequences, value = TRUE)

# 'A{2,3}' = exactly 2 or 3 A's
grep("A{2,3}TGC", sequences, value = TRUE)
```

### Anchors and character classes

```r
genes <- c("BRCA1", "TP53", "EGFR", "MYC", "BRCA2", "KRAS")

# ^ = start of string
grep("^BRC", genes, value = TRUE)

# $ = end of string
grep("1$", genes, value = TRUE)

# [ ] = any character inside brackets
grep("[0-9]", genes, value = TRUE)  # genes with digits
```

### Replacing patterns with gsub()

```r
# Remove all non-alphabetic characters from a messy sequence
messy <- "ATG-CCT!GGT*AA"
clean <- gsub("[^A-Za-z]", "", messy)
clean

# Replace all CG with XY
gsub("CG", "XY", clean)
```

## Final Challenge: Putting It All Together

**The Grand Problem:** You're studying how a new drug affects gene expression in cancer cells.
You performed an RNA-seq experiment with 3 control and 3 treated samples.

Your task — use everything you've learned:

```r
# === STEP 1: Input Data ===
# Create the expression matrix (5 genes × 6 samples)
expression <- matrix(
    c(12, 15, 11, 45, 50, 48,   # GeneA
       8,  9,  7,  6,  5,  7,   # GeneB
      30, 28, 32, 90, 85, 88,   # GeneC
       2,  3,  2,  2,  1,  2,   # GeneD
      55, 60, 58, 20, 22, 18),  # GeneE
    nrow = 5, byrow = TRUE
)
rownames(expression) <- c("GeneA", "GeneB", "GeneC", "GeneD", "GeneE")
colnames(expression) <- c("Ctl1", "Ctl2", "Ctl3", "Trt1", "Trt2", "Trt3")

# === STEP 2: Create a data frame with gene info ===
gene_data <- data.frame(
    Gene = rownames(expression),
    Control_Mean = rowMeans(expression[, 1:3]),
    Treated_Mean = rowMeans(expression[, 4:6])
)

# === STEP 3: Calculate fold change ===
gene_data$FoldChange <- gene_data$Treated_Mean / gene_data$Control_Mean

# === STEP 4: Classify each gene ===
classify_gene <- function(fc) {
    if (fc > 1.5) return("Upregulated")
    else if (fc < 0.67) return("Downregulated")
    else return("No change")
}

gene_data$Status <- sapply(gene_data$FoldChange, classify_gene)

# === STEP 5: Report results ===
cat("=== ANALYSIS RESULTS ===\n\n")
print(gene_data)
cat("\nUpregulated:", sum(gene_data$Status == "Upregulated"), "genes\n")
cat("Downregulated:", sum(gene_data$Status == "Downregulated"), "genes\n")

# === STEP 6: Save ===
write.csv(gene_data, "drug_response_results.csv", row.names = FALSE)
cat("\nResults saved to drug_response_results.csv")
```

| Concept | What you can now do |

| **Data Frames** | Build spreadsheets in R, filter, add columns |
| **Matrices** | Work with expression grids |
| **Factors** | Handle categories (treatment, stage, tissue) |
| **Lists** | Store mixed data types together |
| **If-else** | Classify results automatically |
| **Loops** | Process 20,000 genes in seconds |
| **Functions** | Package analysis into reusable tools |
| **Strings & Regex** | Analyze DNA sequences, find motifs |
| **I/O** | Read collaborator data, save your results |
