## 3. Arithmetic & Relational Operators — Fold Change & Cutoffs

**The problem:** You have raw Ct values for control and treated samples.
How do you calculate fold change? And which genes are truly upregulated?

**The solution:** Arithmetic operators for math; relational operators for comparisons.

### Arithmetic: Calculating 2^(-ΔΔCt)

```r
# Ct values for a reference gene (e.g., GAPDH) and target genes
ct_ref <- 22.1       # GAPDH Ct
ct_target <- 28.5    # Target gene Ct in treated sample
ct_control <- 30.0   # Target gene Ct in control sample

# ΔCt for treated and control
dct_treated <- ct_target - ct_ref
dct_control <- ct_control - ct_ref

# ΔΔCt
ddct <- dct_treated - dct_control

# Fold change = 2^(-ΔΔCt)
fold_change <- 2 ^ (-ddct)
fold_change
```

### Common arithmetic operators

```r
a <- 10
b <- 3

a + b    # addition
a - b    # subtraction
a * b    # multiplication
a / b    # division
a ^ b    # exponentiation (10³ = 1000)
a %% b   # modulo (remainder: 10/3 = 3 remainder 1)
```

### Relational: Which genes pass the cutoff?

```r
fold_changes <- c(1.2, 4.5, 0.3, 8.1, 2.0, 0.9)
gene_names <- c("BRCA1", "TP53", "EGFR", "MYC", "KRAS", "PTEN")

# Which genes have fold change > 2? (upregulated)
fold_changes > 2

# Get the actual gene names that pass the cutoff
gene_names[fold_changes > 2]
```

### All relational operators

```r
values <- c(1, 5, 10, 15, 20)

values > 10    # greater than
values < 10    # less than
values == 10   # equal to (==, not =)
values != 10   # not equal to
values >= 10   # greater than or equal
values <= 10   # less than or equal
```

## 4. Logical Operators — Combining Conditions

**The problem:** You want genes that are BOTH significantly upregulated (fold change > 2)
AND have high expression (mean Ct < 30). How do you combine conditions?

**The solution:** Logical operators — `&` (AND), `|` (OR), `!` (NOT).

```r
fold_changes <- c(1.2, 4.5, 0.3, 8.1, 2.0, 0.9)
ct_values <- c(28.5, 31.2, 27.8, 35.1, 29.6, 33.0)
gene_names <- c("BRCA1", "TP53", "EGFR", "MYC", "KRAS", "PTEN")

# Fold change > 2 AND Ct < 30
condition <- fold_changes > 2 & ct_values < 30
condition

# Which genes satisfy both?
gene_names[condition]
```

### OR and NOT in action

```r
# OR: fold change > 4 OR Ct < 28
gene_names[fold_changes > 4 | ct_values < 28]

# NOT: genes that are NOT upregulated
gene_names[!(fold_changes > 2)]
```

## 9. Decision Making — Classifying Results

**The problem:** Based on fold change and p-value, you want to label each gene as:
- "Upregulated" (FC > 2, p < 0.05)
- "Downregulated" (FC < 0.5, p < 0.05)
- "Not significant" (everything else)

**The solution:** `if-else` statements — your code makes decisions.

```r
# Classify one gene
fc <- 4.5
pval <- 0.003

if (fc > 2 & pval < 0.05) {
    print("Upregulated")
} else if (fc < 0.5 & pval < 0.05) {
    print("Downregulated")
} else {
    print("Not significant")
}
```

## 11. Functions — Build Your Own Tools

**The problem:** You keep calculating 2^(-ΔΔCt) the same way for every experiment.
Copy-paste is error-prone. Can you package this calculation into a reusable tool?

**The solution:** Write your own **function**.

```r
# Write a function to compute fold change from Ct values
fold_change <- function(ct_target, ct_control, ct_ref) {
    dct_treated <- ct_target - ct_ref
    dct_control <- ct_control - ct_ref
    ddct <- dct_treated - dct_control
    fc <- 2 ^ (-ddct)
    return(fc)
}

# Use it
fold_change(ct_target = 28.5, ct_control = 30.0, ct_ref = 22.1)
```

### Functions with default arguments

```r
# Default threshold for significance
classify_gene <- function(fc, pval, fc_cutoff = 2.0, pval_cutoff = 0.05) {
    if (fc > fc_cutoff & pval < pval_cutoff) {
        return("Upregulated")
    } else if (fc < 1/fc_cutoff & pval < pval_cutoff) {
        return("Downregulated")
    } else {
        return("Not significant")
    }
}

# Call with defaults
classify_gene(4.5, 0.003)

# Call with custom cutoff
classify_gene(4.5, 0.003, fc_cutoff = 3.0)
```

### Returning multiple values

```r
# A function that returns summary statistics as a list
gene_summary <- function(values) {
    list(
        mean = mean(values),
        sd = sd(values),
        min = min(values),
        max = max(values)
    )
}

gene_summary(c(1.2, 4.5, 0.3, 8.1, 2.0, 0.9))
```

## 13. Regular Expressions — Finding Patterns in Sequences

**The problem:** You need to find restriction enzyme cut sites in a DNA sequence.
EcoRI cuts at `GAATTC`. How many sites? Where are they?

**The solution:** **Regular expressions** — a mini-language for pattern matching in text.

```r
# Our DNA sequence
seq <- "ATGAATTCGGCCGAATTCGTTAAGAATTCCTG"

# Find the EcoRI sites (GAATTC)
grep("GAATTC", seq, value = TRUE)
```

### Quantifiers — flexible pattern matching

```r
# Sequences to search
sequences <- c("ATGC", "AATGC", "AAATGC", "AAAATGC")

# 'A+' = one or more A's
grep("A+TGC", sequences, value = TRUE)

# 'A{2,3}' = exactly 2 or 3 A's
grep("A{2,3}TGC", sequences, value = TRUE)
```

### Anchors and character classes

```r
genes <- c("BRCA1", "TP53", "EGFR", "MYC", "BRCA2", "KRAS")

# ^ = start of string
grep("^BRC", genes, value = TRUE)

# $ = end of string
grep("1$", genes, value = TRUE)

# [ ] = any character inside brackets
grep("[0-9]", genes, value = TRUE)  # genes with digits
```

### Replacing patterns with gsub()

```r
# Remove all non-alphabetic characters from a messy sequence
messy <- "ATG-CCT!GGT*AA"
clean <- gsub("[^A-Za-z]", "", messy)
clean

# Replace all CG with XY
gsub("CG", "XY", clean)
```

## Final Challenge: Putting It All Together

**The Grand Problem:** You're studying how a new drug affects gene expression in cancer cells.
You performed an RNA-seq experiment with 3 control and 3 treated samples.

Your task — use everything you've learned:

```r
# === STEP 1: Input Data ===
# Create the expression matrix (5 genes × 6 samples)
expression <- matrix(
    c(12, 15, 11, 45, 50, 48,   # GeneA
       8,  9,  7,  6,  5,  7,   # GeneB
      30, 28, 32, 90, 85, 88,   # GeneC
       2,  3,  2,  2,  1,  2,   # GeneD
      55, 60, 58, 20, 22, 18),  # GeneE
    nrow = 5, byrow = TRUE
)
rownames(expression) <- c("GeneA", "GeneB", "GeneC", "GeneD", "GeneE")
colnames(expression) <- c("Ctl1", "Ctl2", "Ctl3", "Trt1", "Trt2", "Trt3")

# === STEP 2: Create a data frame with gene info ===
gene_data <- data.frame(
    Gene = rownames(expression),
    Control_Mean = rowMeans(expression[, 1:3]),
    Treated_Mean = rowMeans(expression[, 4:6])
)

# === STEP 3: Calculate fold change ===
gene_data$FoldChange <- gene_data$Treated_Mean / gene_data$Control_Mean

# === STEP 4: Classify each gene ===
classify_gene <- function(fc) {
    if (fc > 1.5) return("Upregulated")
    else if (fc < 0.67) return("Downregulated")
    else return("No change")
}

gene_data$Status <- sapply(gene_data$FoldChange, classify_gene)

# === STEP 5: Report results ===
cat("=== ANALYSIS RESULTS ===\n\n")
print(gene_data)
cat("\nUpregulated:", sum(gene_data$Status == "Upregulated"), "genes\n")
cat("Downregulated:", sum(gene_data$Status == "Downregulated"), "genes\n")

# === STEP 6: Save ===
write.csv(gene_data, "drug_response_results.csv", row.names = FALSE)
cat("\nResults saved to drug_response_results.csv")
```

### The Universal 6-Step Workflow for Statistical Problems

Whenever you encounter a statistical research question in the lab or clinic, follow this **6-step framework** (PDF Slide 20):

1. **State the Hypotheses**:
   * **Null Hypothesis ($H_0$)**: There is NO difference, NO effect, or NO association. Any observed pattern is due purely to random sampling chance.
   * **Alternative Hypothesis ($H_1$)**: There IS a real, significant difference, effect, or association.
2. **Select the Appropriate Statistical Test**:
   * Identify your data type: *Continuous measurement* vs *Categorical/counts* vs *Ranks/ordinal*.
   * Check your experimental design: *1 group vs benchmark*, *2 independent groups*, *paired before/after*, *more than 2 groups*, or *correlation/prediction*.
3. **Verify Assumptions**:
   * **Parametric tests** (like $t$-tests and ANOVA) assume continuous data, random independent sampling, normal distribution (bell curve), and homoscedasticity (equal variance).
   * **Nonparametric tests** (like $\chi^2$, Mann-Whitney $U$, Wilcoxon signed-rank, Kruskal-Wallis) make no distribution assumptions and work wonderfully for qualitative data, ranks, or small sample sizes ($n < 30$).
4. **Calculate the Test Statistic**:
   * Compute sums, means, standard deviations, and standard errors, then apply the formula for your test statistic ($t$, $\chi^2$, $U$, $W$, $r$, $F$, or $H$).
5. **Determine Degrees of Freedom ($df$) & Critical Cutoff**:
   * At significance level $\alpha = 0.05$ (the top 5% threshold of extreme results), locate the critical value from statistical tables.
6. **Draw Scientific Inference**:
   * If your computed test statistic exceeds the critical boundary (or $p$-value $< \alpha = 0.05$), **reject $H_0$** and conclude the result is statistically significant!
   * If not, **retain $H_0$** (fail to reject).

## 16. Independent Samples $t$-test: Comparing Two Independent Groups

### Step 1: Problem Statement & Requirements (PDF Slide 12)
Researchers measured the serum **C-reactive protein (CRP)** concentration (in mg/L)—a key biomarker of systemic inflammation—in two distinct groups of individuals:
* **Group 1 (Patients with inflammatory disease, $n_1 = 12$):** `4.9, 5.2, 4.7, 5.3, 3.9, 5.4, 4.6, 4.9, 4.8, 5.1, 4.2, 4.8`
* **Group 2 (Healthy controls, $n_2 = 10$):** `3.1, 3.7, 3.5, 4.0, 3.3, 3.4, 3.3, 3.2, 3.0, 3.2`

**Research Question:** Is there a statistically significant difference in the mean serum CRP concentration between inflammatory disease patients and healthy controls?

**Formulate Hypotheses:**
* **Null Hypothesis ($H_0$):** $\mu_1 = \mu_2$ (There is no difference in mean serum CRP between patients and controls; any difference is due to chance).
* **Alternative Hypothesis ($H_1$):** $\mu_1 \neq \mu_2$ (Mean serum CRP differs significantly between patients and controls).
* **Significance Level:** $\alpha = 0.05$.
* **Given Critical Score from Table (PDF Slide 12):** $t_{0.05}(20) = 2.086$.

---

### Step 2: How to Solve as in the PDF (PDF Slide 13)
1. **Sample Sizes & Group Means:**
   $$\bar{X}_1 = \frac{\sum X_1}{n_1} = \frac{57.8}{12} = 4.82\text{ mg/L}, \quad \bar{X}_2 = \frac{\sum X_2}{n_2} = \frac{33.7}{10} = 3.37\text{ mg/L}$$

2. **Sum of Squared Deviations:**
   $$\sum(X_1 - \bar{X}_1)^2 = 2.0968, \quad \sum(X_2 - \bar{X}_2)^2 = 0.8010$$

3. **Pooled Standard Deviation ($\hat{s}$):**
   $$\hat{s} = \sqrt{\frac{\sum(X_1 - \bar{X}_1)^2 + \sum(X_2 - \bar{X}_2)^2}{n_1 + n_2 - 2}} = \sqrt{\frac{2.0968 + 0.8010}{12 + 10 - 2}} = 0.386$$

4. **Standard Error of the Difference ($S_{\bar{X}_1 - \bar{X}_2}$):**
   $$S_{\bar{X}_1 - \bar{X}_2} = \hat{s} \times \sqrt{\frac{n_1 + n_2}{n_1 \times n_2}} = 0.386 \times \sqrt{\frac{12 + 10}{120}} = 0.1653$$

5. **Test Statistic ($t$) & Degrees of Freedom ($df$):**
   $$t = \frac{\bar{X}_1 - \bar{X}_2}{S_{\bar{X}_1 - \bar{X}_2}} = \frac{4.82 - 3.37}{0.1653} = 8.772, \quad df = 12 + 10 - 2 = 20$$


```r
# ==============================================================================
# Step 3A: MINIMUM CODE (Explain this to students!)
# Just define data and run t.test() in ONE line!
# ==============================================================================

crp_patients <- c(4.9, 5.2, 4.7, 5.3, 3.9, 5.4, 4.6, 4.9, 4.8, 5.1, 4.2, 4.8)
crp_controls <- c(3.1, 3.7, 3.5, 4.0, 3.3, 3.4, 3.3, 3.2, 3.0, 3.2)

# The entire test in a single command:
t.test(crp_patients, crp_controls, var.equal = TRUE)
```

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Confidence Intervals)
# ==============================================================================

t_test_crp <- t.test(crp_patients, crp_controls, var.equal = TRUE)

cat("=== Independent Samples t-test Results ===\n")
cat("Patient Mean (X1-bar)   :", round(mean(crp_patients), 2), "mg/L\n")
cat("Control Mean (X2-bar)   :", round(mean(crp_controls), 2), "mg/L\n")
cat("Calculated t-statistic  :", round(t_test_crp$statistic, 3), "\n")
cat("Degrees of Freedom (df) :", t_test_crp$parameter, "\n")
cat("p-value                 :", format.pval(t_test_crp$p.value, eps = 0.0001), "\n")
cat("95% Confidence Interval :", round(t_test_crp$conf.int[1], 2), "to", round(t_test_crp$conf.int[2], 2), "mg/L\n")
```

### Step 4: Data Visualization — Teaching the Box Plot (`boxplot()`)

*(Reference: [Bakshi Lab Guide: Box Plots](https://www.bakshilab.org/blog/basic_r_2#h.d-box-plots))*

**Why a Box Plot?**
A box plot displays the 5-number summary (Minimum, $Q_1$, Median $Q_2$, $Q_3$, Maximum). It is the premier plot to compare distributions of two groups.

### Step 5: Inference of the Results (PDF Slide 14)

* **Statistical Comparison:** 
  $$\text{Computed } t = 8.876 \text{ (manual } 8.772\text{)} > \text{Critical } t_{0.05}(20) = 2.086$$
  $$p\text{-value} = 2.26 \times 10^{-8} \ll 0.05$$
* **Decision on $H_0$:** Reject the Null Hypothesis ($H_0$).
* **Biological Takeaway:** There is a **statistically significant difference** between group means. Patients with inflammatory disease have substantially elevated serum CRP levels (mean $4.82$ mg/L) compared to healthy controls (mean $3.37$ mg/L), confirming CRP as a biomarker of systemic inflammation.

## 17. Paired $t$-test: Dependent Before-and-After Measurements

### Step 1: Problem Statement & Requirements (PDF Slide 15)
A physician measured the **diastolic blood pressure (mmHg)** of $10$ hypertensive patients **Before** ($X_1$) and **4 weeks After** ($X_2$) initiating an antihypertensive therapy:

**Research Question:** Is there a statistically significant difference in mean diastolic blood pressure before and after treatment?

**Formulate Hypotheses:**
* **Null Hypothesis ($H_0$):** $\mu_D = 0$ (Treatment causes no change; mean difference before vs after is zero).
* **Alternative Hypothesis ($H_1$):** $\mu_D \neq 0$ (Treatment significantly changes diastolic blood pressure).
* **Significance Level:** $\alpha = 0.05$.
* **Given Critical Score from Table (PDF Slide 15):** $t_{0.05}(9) = 2.262$.

### Step 2: How to Solve as in the PDF (PDF Slide 16)
1. **Pairwise Differences ($D = X_1 - X_2$):**

   $$t = \frac{\bar{D}}{s_{\bar{D}}} = \frac{7.89}{0.5015} = 15.733, \quad df = n - 1 = 9$$

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Diagnostics)
# ==============================================================================

t_paired <- t.test(bp_before, bp_after, paired = TRUE)

cat("=== Paired t-test Results ===\n")
cat("Mean Reduction (D-bar)  :", round(mean(bp_before - bp_after), 2), "mmHg\n")
cat("Calculated t-statistic  :", round(t_paired$statistic, 3), "\n")
cat("Degrees of Freedom (df) :", t_paired$parameter, "\n")
cat("p-value                 :", format.pval(t_paired$p.value, eps = 0.0001), "\n")
cat("95% CI of Reduction     :", round(t_paired$conf.int[1], 2), "to", round(t_paired$conf.int[2], 2), "mmHg\n")
```

## 18. One-Sample $t$-test: Benchmarking Against a Safety Standard

The food safety regulatory standard states that the maximum safe average concentration is **$\mu_0 = 0.5$ ppm**.

**Research Question:** Does the swordfish population significantly exceed the safety threshold?

**Formulate Hypotheses:**
* **Null Hypothesis ($H_0$):** $\mu = 0.5$ ppm (Population mean equals safety threshold).
* **Alternative Hypothesis ($H_1$):** $\mu > 0.5$ ppm (Population mean significantly exceeds threshold).

### Step 2: How to Solve as in the PDF (PDF Slide 18)
1. **Sample Mean ($\bar{X}$) and Sample SD ($s$):**
   $$\bar{X} = 1.338\text{ ppm}, \quad s = 0.248\text{ ppm}$$
2. **Standard Error of Mean ($SE$):**
   $$SE = \frac{s}{\sqrt{n}} = \frac{0.248}{\sqrt{15}} = 0.0641\text{ ppm}$$
3. **Test Statistic ($t$) & Degrees of Freedom ($df$):**
   $$t = \frac{\bar{X} - \mu_0}{SE} = \frac{1.338 - 0.500}{0.0641} = 13.07, \quad df = 15 - 1 = 14$$

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Directional Test)
# ==============================================================================

t_one <- t.test(mercury, mu = 0.5, alternative = "greater")

cat("=== One-Sample t-test Results ===\n")
cat("Sample Mean (X-bar)     :", round(mean(mercury), 3), "ppm\n")
cat("Regulatory Cutoff (mu0) : 0.50 ppm\n")
cat("Calculated t-statistic  :", round(t_one$statistic, 3), "\n")
cat("Degrees of Freedom (df) :", t_one$parameter, "\n")
cat("p-value                 :", format.pval(t_one$p.value, eps = 0.0001), "\n")
```

## 19. Chi-Square ($\chi^2$) Goodness-of-Fit Test: Mendelian Genetics

### Step 1: Problem Statement & Requirements (PDF Slide 20)
In a plant breeding experiment, flower colors of $N = 929$ progeny were recorded:
* **Red flower:** $705$ plants
* **White flower:** $224$ plants

**Research Question:** Justify whether the Mendelian monohybrid $3:1$ ratio hypothesis prevails.

**Formulate Hypotheses (PDF Slide 21):**
* **Null Hypothesis ($H_0$):** Results conform to the Mendelian $3:1$ ratio (differences are due purely to chance).
* **Alternative Hypothesis ($H_1$):** Results deviate significantly from the $3:1$ ratio.
* **Significance Level:** $\alpha = 0.05, \quad df = 2 - 1 = 1$.
* **Critical Score (PDF Slide 24):** $\chi^2_{0.05}(1) = 3.841$.

---

### Step 2: How to Solve as in the PDF (PDF Slide 22-24)
1. **Expected Counts:**
   $$E_{\text{red}} = 929 \times 3/4 = 697, \quad E_{\text{white}} = 929 \times 1/4 = 232$$
2. **Chi-Square Formula (PDF Slide 23):**
   $$\chi^2 = \sum \frac{(O - E)^2}{E} = \frac{(705 - 697)^2}{697} + \frac{(224 - 232)^2}{232} = 0.09 + 0.27 = 0.36$$


```r
# ==============================================================================
# Step 3A: MINIMUM CODE (Explain this to students!)
# Just chisq.test() with expected probabilities in ONE line!
# ==============================================================================

# The entire test in a single command:
chisq.test(c(705, 224), p = c(3/4, 1/4))
```

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Observed vs Expected)
# ==============================================================================

flower_obs <- c(Red = 705, White = 224)
chi_res <- chisq.test(flower_obs, p = c(3/4, 1/4), correct = FALSE)

cat("=== Chi-Square Goodness-of-Fit Results ===\n")
cat("Calculated Chi-Square :", round(chi_res$statistic, 3), "\n")
cat("Degrees of Freedom    :", chi_res$parameter, "\n")
cat("p-value               :", round(chi_res$p.value, 4), "\n")
cat("Critical Cutoff (df=1): 3.841\n")
```

### Step 4: Data Visualization — Teaching the Grouped Bar Plot (`barplot()`)

*(Reference: [Bakshi Lab Guide: Bar Plots](https://www.bakshilab.org/blog/basic_r_2#h.c-bar-plots))*

**Why a Bar Plot?**

## 20. Mann-Whitney $U$ Test: Nonparametric Two-Group Comparison

### Step 1: Problem Statement & Requirements (PDF Slide 27)
A public health researcher investigated whether participation in a nutrition intervention program improves **metabolic scores** ($n = 8$ per group):
* **Non-participants ($n_1 = 8$):** `10.2, 11.5, 9.8, 12.1, 10.7, 11.0, 9.9, 10.5`
* **Participants ($n_2 = 8$):** `12.8, 13.4, 14.1, 13.0, 15.2, 14.5, 13.8, 14.9`

**Hypotheses:**
* **$H_0$:** No difference in metabolic scores between groups.
* **$H_1$:** Participants have significantly higher scores.
* **Given Critical Score (PDF Slide 27):** $U_{0.05}(8, 8) = 13$.

---

### Step 2: How to Solve as in the PDF (PDF Slide 28-29)
1. Combine and rank all 16 scores: Non-participants hold ranks 1-8 ($R_1 = 36$); Participants hold ranks 9-16 ($R_2 = 100$).
2. $U_1 = 64, U_2 = 0 \implies U = \min(64, 0) = 0$.
3. **CRITICAL RULE (PDF Slide 29):** In Mann-Whitney $U$, reject $H_0$ if Computed $U \le \text{Critical } U$! ($0 \le 13 \implies$ Reject $H_0$).


```r
# ==============================================================================
# Step 3A: MINIMUM CODE (Explain this to students!)
# Just wilcox.test() in ONE line!
# ==============================================================================

non_part <- c(10.2, 11.5, 9.8, 12.1, 10.7, 11.0, 9.9, 10.5)
part     <- c(12.8, 13.4, 14.1, 13.0, 15.2, 14.5, 13.8, 14.9)

# The entire Mann-Whitney test in a single command:
wilcox.test(non_part, part)
```

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Diagnostics)
# ==============================================================================

mw_res <- wilcox.test(non_part, part, exact = FALSE)

cat("=== Mann-Whitney U Test Results ===\n")
cat("Calculated W / U  :", mw_res$statistic, "\n")
cat("p-value           :", format.pval(mw_res$p.value, eps = 0.0001), "\n")
cat("Critical Cutoff   : 13 (Reject H0 if U <= 13)\n")
```

## 21. Wilcoxon Signed-Rank Test: Nonparametric Paired Comparison

### Step 1: Problem Statement & Requirements (PDF Slide 30)
$TP53$ gene expression fold change was measured in $11$ cell cultures **Before** ($X_1$) and **After** ($X_2$) CRISPR treatment:
* **Before:** `15.5, 12.7, 14.8, 16.7, 20.1, 22.0, 20.2, 18.1, 17.6, 17.4, 19.1`
* **After:** `21.2, 20.1, 17.2, 22.7, 20.0, 19.8, 19.8, 18.8, 17.9, 24.3, 18.6`

**Hypotheses:**
* **$H_0$:** CRISPR treatment causes no change in $TP53$ expression.
* **$H_1$:** CRISPR treatment significantly changes $TP53$ expression.
* **Given Critical Score (PDF Slide 30):** $W_{0.05}(11) = 11$.

---

### Step 2: How to Solve as in the PDF (PDF Slide 31-32)
1. Pairwise differences: 7 positive, 4 negative.
2. Sum of ranks: $W^+ = 52, W^- = 14 \implies W = \min(52, 14) = 14$.
3. **DECISION RULE:** Reject $H_0$ only if Computed $W \le \text{Critical } W$ ($11$).


```r
# ==============================================================================
# Step 3A: MINIMUM CODE (Explain this to students!)
# Just wilcox.test() with paired = TRUE in ONE line!
# ==============================================================================

tp53_before <- c(15.5, 12.7, 14.8, 16.7, 20.1, 22.0, 20.2, 18.1, 17.6, 17.4, 19.1)
tp53_after  <- c(21.2, 20.1, 17.2, 22.7, 20.0, 19.8, 19.8, 18.8, 17.9, 24.3, 18.6)

# The entire paired test in a single command:
wilcox.test(tp53_after, tp53_before, paired = TRUE)
```

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Diagnostics)
# ==============================================================================

ws_res <- wilcox.test(tp53_after, tp53_before, paired = TRUE, exact = FALSE)

cat("=== Wilcoxon Signed-Rank Test Results ===\n")
cat("R Test Statistic (V = W+)   :", ws_res$statistic, "\n")
cat("Manual Statistic W = min(W+,W-): 14\n")
cat("p-value                     :", round(ws_res$p.value, 4), "\n")
cat("Critical Cutoff             : 11 (Reject H0 only if W <= 11)\n")
```

* **$H_1$:** $\rho \neq 0$ (Significant linear correlation).
* **Given Critical Score (PDF Slide 34):** $t_{0.05}(8) = 2.306$.

---

### Step 2: How to Solve as in the PDF (PDF Slide 35)
1. $r = \frac{10(2707.1) - (32.4)(825.4)}{\sqrt{[1077.2 - 1049.76][687103.6 - 681285.16]}} = +0.95$
2. $t = \frac{r}{S_r} = \frac{0.95}{0.110} = 8.636, \quad df = 8$
3. Coefficient of determination: $r^2 = (0.95)^2 = 0.9025$ ($90\%$ of variance explained).


```r
# ==============================================================================
# Step 3A: MINIMUM CODE (Explain this to students!)
# Just cor.test() in ONE line!
# ==============================================================================

o2_consumption       <- c(3.0, 3.8, 3.5, 3.0, 3.8, 2.5, 2.7, 2.9, 3.7, 3.5)
tracheal_ventilation <- c(84.2, 90.0, 88.5, 76.0, 90.1, 71.5, 72.5, 74.0, 90.0, 88.5)

# The entire correlation test in a single command:
cor.test(o2_consumption, tracheal_ventilation)
```

* **$H_0$:** Grip strength distributions are identical across all three therapies.
* **$H_1$:** At least one therapy produces a significantly different grip strength distribution.
* **Given Critical Score (PDF Slide 47):** $\chi^2_{0.05}(2) = 5.991$.

---

### Step 2: How to Solve as in the PDF (PDF Slide 47)
1. Group rank sums: $R_A = 15, R_B = 40, R_C = 65$
2. $H = \frac{12}{15(16)} [45 + 320 + 845] - 48 = 12.5, \quad df = 2$


```r
# ==============================================================================
# Step 3A: MINIMUM CODE (Explain this to students!)
# Just kruskal.test() in ONE line!
# ==============================================================================

rehab_df <- data.frame(
  grip = c(18, 20, 22, 19, 21,
           25, 27, 29, 26, 28,
           32, 34, 36, 33, 35),
  therapy = factor(rep(c("A", "B", "C"), each = 5))
)

# The entire Kruskal-Wallis test in a single command:
kruskal.test(grip ~ therapy, data = rehab_df)
```

```r
# ==============================================================================
# Step 3B: DETAILED CODE (Formatted Output & Diagnostics)
# ==============================================================================

kw_res <- kruskal.test(grip ~ therapy, data = rehab_df)

cat("=== Kruskal-Wallis Test Results ===\n")
cat("Kruskal-Wallis H (chi-sq) :", round(kw_res$statistic, 3), "\n")
cat("Degrees of Freedom (df)   :", kw_res$parameter, "\n")
cat("p-value                   :", round(kw_res$p.value, 5), "\n")
cat("Critical Cutoff           : 5.991\n")
```

### Step 4: Data Visualization — Teaching Comparative Ranked Group Plots

*(Reference: [Bakshi Lab Guide: Box Plots](https://www.bakshilab.org/blog/basic_r_2#h.d-box-plots))*


```r
# ==============================================================================
# Step 4A: MINIMUM PLOT (Explain this to students!)
# Just boxplot() in ONE line!
# ==============================================================================

boxplot(grip ~ therapy, data = rehab_df)
```

```r
# ==============================================================================
# Step 4B: DETAILED PLOT (Publication-Ready Box Plot)
# ==============================================================================

par(mar = c(5, 5, 4, 2), bg = "white")

boxplot(grip ~ therapy, data = rehab_df,
        col = c("#7FC97F", "#BEAED4", "#FDC086"), border = "gray20", lwd = 1.5,
        main = "Grip Strength Recovery Across Rehabilitation Therapies",
        ylab = "Hand Grip Strength (kg)", xlab = "Therapy Regimen",
        ylim = c(15, 40), las = 1)

grid(nx = NA, ny = NULL, col = "gray85", lty = "dotted")
stripchart(grip ~ therapy, data = rehab_df, vertical = TRUE, method = "jitter",
           add = TRUE, pch = 16, col = "black", cex = 1.3)
text(2, 39, "*** p = 0.0019 (H = 12.5 > 5.991)", font = 2, col = "firebrick")
```

### Step 5: Inference of the Results (PDF Slide 47)

* **Statistical Comparison:** 
  $$\text{Computed } H = 12.5 > \text{Critical } \chi^2_{0.05}(2) = 5.991$$
  $$p\text{-value} = 0.00193 < 0.05$$
* **Decision on $H_0$:** Reject the Null Hypothesis ($H_0$).

## 27. Heatmap Visualization: Multi-Variable & Omics Data

*(Reference: [Bakshi Lab Guide: Heatmap Visualization](https://www.bakshilab.org/blog/basic_r_2#h.28nnusbuqmzy))*

**Why Heatmaps in Biology?**
A heatmap visualizes a 2D numerical matrix where colors represent values. It is standard for RNA-seq and multi-variable phenotyping.


```r
# ==============================================================================
# Step 1A: MINIMUM HEATMAP (Explain this to students!)
# Just heatmap() in ONE line!
# ==============================================================================

expr_matrix <- matrix(
  c(12.5, 13.1, 24.5, 26.2,  # TP53
     8.2,  7.9, 18.4, 19.8,  # CDKN1A
    25.1, 26.4, 11.2,  9.8,  # MYC
    18.0, 17.8, 18.2, 17.9,  # ACTB
     6.1,  6.8, 21.0, 23.5), # BAX
  nrow = 5, byrow = TRUE,
  dimnames = list(c("TP53", "CDKN1A", "MYC", "ACTB", "BAX"),
                  c("Ctrl_1", "Ctrl_2", "Drug_1", "Drug_2"))
)

# The entire heatmap in a single command:
heatmap(expr_matrix)
```

```r
# ==============================================================================
# Step 1B: DETAILED CLUSTERED HEATMAP (Using pheatmap with Metadata)
# ==============================================================================

library(pheatmap)

sample_meta <- data.frame(
  Treatment = c("Control", "Control", "Drug-Treated", "Drug-Treated"),
  row.names = colnames(expr_matrix)
)

pheatmap(expr_matrix,
         scale = "row",
         annotation_col = sample_meta,
         color = colorRampPalette(c("#313695", "#FFFFBF", "#A50026"))(100),
         display_numbers = TRUE, number_format = "%.1f",
         main = "Clustered Gene Expression Heatmap (pheatmap)",
         cellwidth = 55, cellheight = 32)
```

### Biological Inference of Heatmaps
* **Sample Clustering:** `Ctrl_1` and `Ctrl_2` cluster together, while `Drug_1` and `Drug_2` cluster together.
* **Gene Modules:** Pro-apoptotic genes (*TP53*, *BAX*, *CDKN1A*) are up-regulated after drug treatment, while *MYC* is repressed. Housekeeping gene *ACTB* remains unchanged.
